# **03 - Submission**
Memuat model final, memprediksi `test.csv`, dan membuat `submission.csv` sesuai format `sample_submission.csv`.

```python
import warnings
warnings.filterwarnings("ignore")

import sys, os, pickle
sys.path.append("..")

import pandas as pd
from scripts.predict import make_submission
```

# **1. Read Dataset & Model**

```python
test = pd.read_csv("../data/test.csv")
sample = pd.read_csv("../data/sample_submission.csv")
with open("../models/pipeline_lgbm_structural_damage.pkl", "rb") as f:
    model = pickle.load(f)
print(test.shape, sample.shape)
sample.head()
```

# **2. Predict & Build Submission**

```python
submission = make_submission(model, test, sample, numeric=True)  # angka 1-5 (evaluator Kaggle gagal membaca teks "Grade 3")
submission.head()
```

# **3. Validasi Format**

```python
assert list(submission.columns) == list(sample.columns)
assert len(submission) == len(sample)
assert (submission["building_id"].values == sample["building_id"].values).all()
assert set(submission["damage_grade"]) <= {1, 2, 3, 4, 5}
print("Format submission valid")
submission["damage_grade"].value_counts(normalize=True).sort_index().round(3)
```

# **4. Simpan**

```python
os.makedirs("../submissions", exist_ok=True)
submission.to_csv("../submissions/submission.csv", index=False)
print("Tersimpan: ../submissions/submission.csv")
```
